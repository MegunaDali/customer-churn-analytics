```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

path = Path("data/raw/customer_churn.csv")

if not path.exists():
    path = Path("../data/raw/customer_churn.csv")

df = pd.read_csv(path)

df.columns = df.columns.str.strip().str.replace(
    r"\s+", " ", regex=True
)

df.head()
```

```python
df["Call Failure"].value_counts().sort_index()
```

```python
df.groupby("Churn")["Call Failure"].agg(
    customers="size",
    average_failures="mean",
    median_failures="median",
    maximum_failures="max"
).round(2)
```

```python
df["failure_group"] = pd.cut(
    df["Call Failure"],
    bins=[-1, 0, 5, 10, 20, float("inf")],
    labels=["0", "1-5", "6-10", "11-20", "21+"]
)

df.groupby("failure_group", observed=True)["Churn"].agg(
    customers="count",
    churn_rate="mean"
)
```

```python
import matplotlib.pyplot as plt

df["Call Failure"].hist(
    bins=37,
    figsize=(9, 4),
    edgecolor="white"
)

plt.title("Distribution of Call Failures")
plt.xlabel("Number of Failed Calls")
plt.ylabel("Number of Customers")
plt.show()
```

```python
df.groupby("failure_group", observed=True).agg(
    customers=("Churn", "size"),
    churn_rate=("Churn", "mean"),
    avg_call_frequency=("Frequency of use", "mean"),
    avg_seconds_of_use=("Seconds of Use", "mean")
).round(2)
```

```python
pd.crosstab(
    df["Complains"],
    df["Churn"],
    normalize="index"
) * 100
```

```python
df.groupby(["Complains", "Churn"])["Call Failure"].agg(
    ["count", "mean", "median"]
)
```

```python
df["failure_group"] = pd.cut(
    df["Call Failure"],
    bins=[-1, 0, 5, 10, 20, float("inf")]
)

df.groupby(
    ["Complains", "failure_group"],
    observed=True
)["Churn"].agg(
    customers="count",
    churn_rate="mean"
)
```
